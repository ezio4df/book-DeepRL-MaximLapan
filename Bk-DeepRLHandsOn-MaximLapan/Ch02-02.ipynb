```python
env = gym.make("CartPole-v1")
total_reward = 0.0
total_steps = 0
obs, _ = env.reset()

while True:
    action = env.action_space.sample()
    obs, reward, is_done, is_trunc, _ = env.step(action)

    total_reward += reward
    total_steps += 1

    if is_done:
        break

print(f"Episode finished in {total_steps} steps, total reward {total_reward:.2f}")
```

```python
import gymnasium as gym
import random
```

```python
class RandomActionWrapper(gym.ActionWrapper):
    def __init__(self, env: gym.Env, epsilon: float = 1.0):
        super(RandomActionWrapper, self).__init__(env)
        self.epsilon = epsilon

    def action(self, action: gym.core.WrapperActType) -> gym.core.WrapperActType:
        if random.random() < self.epsilon:
            action = self.env.action_space.sample()
            print(f"Random action: {action}")
            return action

        return action
```

```python
env = RandomActionWrapper(gym.make("CartPole-v1"))
obs = env.reset()
total_reward = 0.0
total_steps = 0

while True:
    obs, reward, is_done, _, _ = env.step(0)
    total_reward += reward
    total_steps += 1
    if is_done:
        break

print(f"Episode finished in {total_steps} steps, total reward {total_reward:.2f}")
```
