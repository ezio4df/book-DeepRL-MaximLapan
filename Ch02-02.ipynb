```python
import gymnasium as gym
from IPython.core.display_functions import display
```

```python
e = gym.make("CartPole-v1")
obs, info = e.reset()

display(obs)
display(info)
```

```python
display(e.action_space)
display(e.observation_space)
```

```python
display(e.action_space.sample())
display(e.observation_space.sample())
```

```python
e.step(0)
```

```python
env = gym.make("CartPole-v1")
total_reward = 0.0
total_steps = 0
obs, _ = env.reset()

while True:
    action = e.action_space.sample()
    obs, reward, is_done, is_trunc, _ = env.step(action)

    total_reward += reward
    total_steps += 1

    if is_done:
        break

print(f"Episode finished in {total_steps} steps, total reward {total_reward:.2f}")
```
